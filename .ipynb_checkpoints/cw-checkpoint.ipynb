```python
def locate_rh_sign(sign_image, scene_image):
    # sign_image: The image of the road sign you need to detect.
    # scene_image: The image in which you need to locate the road sign.
    
        # Load images
    sign_image = cv2.imread(sign_image_path, cv2.IMREAD_COLOR)
    scene_image = cv2.imread(scene_image_path, cv2.IMREAD_COLOR)
    
    # Convert to grayscale
    gray_sign = cv2.cvtColor(sign_image, cv2.COLOR_BGR2GRAY)
    gray_scene = cv2.cvtColor(scene_image, cv2.COLOR_BGR2GRAY)
    
    # Enhance contrast and reduce noise
    gray_sign = cv2.equalizeHist(gray_sign)
    gray_scene = cv2.equalizeHist(gray_scene)
    
    # Use ORB (Oriented FAST and Rotated BRIEF) to detect keypoints and descriptors
    orb = cv2.ORB_create()
    keypoints1, descriptors1 = orb.detectAndCompute(gray_sign, None)
    keypoints2, descriptors2 = orb.detectAndCompute(gray_scene, None)
    
    # Match features using BFMatcher
    bf = cv2.BFMatcher(cv2.NORM_HAMMING, crossCheck=True)
    matches = bf.match(descriptors1, descriptors2)
    
    # Sort matches by distance
    matches = sorted(matches, key=lambda x: x.distance)
    
    # Filter for good matches
    good_matches = matches[:int(len(matches) * 0.2)]  # Use top 20%
    
    # Draw matches (optional visualization)
    match_img = cv2.drawMatches(sign_image, keypoints1, scene_image, keypoints2, good_matches, None, flags=cv2.DrawMatchesFlags_NOT_DRAW_SINGLE_POINTS)
    plt.imshow(cv2.cvtColor(match_img, cv2.COLOR_BGR2RGB))
    plt.title("Feature Matches")
    plt.show()
    
    # Extract locations of good matches
    src_pts = np.float32([keypoints1[m.queryIdx].pt for m in good_matches]).reshape(-1, 1, 2)
    dst_pts = np.float32([keypoints2[m.trainIdx].pt for m in good_matches]).reshape(-1, 1, 2)
    
    # Find homography matrix
    M, mask = cv2.findHomography(src_pts, dst_pts, cv2.RANSAC, 5.0)
    matches_mask = mask.ravel().tolist()
    
    # Get bounding box for the sign image
    h, w = sign_image.shape[:2]
    pts = np.float32([[0, 0], [0, h], [w, h], [w, 0]]).reshape(-1, 1, 2)
    dst = cv2.perspectiveTransform(pts, M)
    
    # Draw bounding box on the scene image
    result_img = scene_image.copy()
    if dst is not None:
        dst = np.int32(dst)
        cv2.polylines(result_img, [dst], True, (0, 255, 0), 3, cv2.LINE_AA)
    
    # Show result
    plt.imshow(cv2.cvtColor(result_img, cv2.COLOR_BGR2RGB))
    plt.title("Detected Road Sign")
    plt.show()
    
    return result_img
```
